We scored each of the 5 bookmakers that we chose, based on the historical Premier League data for the seasons 21/22, 22/23 and 23/24, using the Brier score function:

Looking at the data, rather expectedly, the scores of the bookmakers were similar for every season. However, the average across each season varied slightly:

| Season | Average |
| :--- | :---: |
| 21-22 | 0.5540 |
| 22-23 | 0.5736 |
| 23-24 | 0.5312 |

This could be due to various things, such as how predictable the results were in each season. Draws are the hardest result to predict. The bookmakers gave a draw only about a 23% chance on average, so when one happens, most of the probability was placed on the wrong outcomes and the squared differences are large. In our data, a draw scored about 0.91 on average, compared with about 0.37 to 0.41 for a home win and 0.51 to 0.58 for an away win. Upsets, where the favourite does not win, push the score up in the same way. Seasons with fewer of these score lower: 23-24 had the fewest draws (21% of matches) and the favourite won most often (about 60%), and it had the lowest average score. In 22-23 the favourite won least often (about 56%), and it had the highest. All of the bookmakers are affected in the same way, which is why their scores move together from season to season. With only 380 matches per season, chance also plays a part. At least there is consistency.

Again, this is fairly random. However, we can see from the average across the 3 seasons that Pinnacle scores the lowest and William Hill the highest.

The Brier score gives a higher score when the probabilities stray further away from the real events, due to the way the function works:

1. We convert the odds to probabilities, normalising them so that they sum to 1.
2. We create a vector in the format H, D, A, where each element is 1 if that result happened and 0 otherwise.
3. We calculate the mean squared error: for each match, we take the difference between each probability and the outcome, square it and sum over the three outcomes, then average over all matches.

Hence, the closer a probability is to what actually happened (near 1 for the result that happened, and near 0 for the results that did not), the smaller the squared difference and the lower the score. A low score therefore means the probabilities were close to the real events.

Overall, Pinnacle is the most accurate, with the lowest average score of 0.5527, and William Hill the least accurate, with the highest at 0.5533. However, it is worth noting that the difference is very small (0.0006). Pinnacle also had the lowest margin of the five bookmakers, at about 2.6% to 3.5% per season, compared with about 5% to 7.6% for the others. Pinnacle and VC Bet are identical to 4 d.p., so they are effectively tied. We could not find any source saying that they use the same odds maker, and when we compared their odds match by match, none were identical (they differed by about 0.2 on average). It is more likely that their scores are close because bookmakers' odds track each other closely once the margin is removed.

We chose these seasons because they had a full amount of data, bar 2 rows in the latest season. Each season has 38 matchweeks with 10 games in each, giving 380 matches: each of the 20 teams plays the other 19 teams both home and away, so 2 × 19 = 38 matches per team.
